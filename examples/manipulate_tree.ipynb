# Manipulating a YAML tree in memory

This notebook reads a YAML file and then manipulates the resulting tree
structure in memory: inspecting nodes, accessing sequence and map elements,
adding new elements, and writing the edited tree back to a file.

`import palsparserpy as pp` brings the whole public API into scope under a short
name, which is how the functions are called throughout this notebook. Every tree
operation is also a method on the node itself, so `pp.add_map(facility)` and
`facility.add_map()` are the same call written two ways.

The package has to be importable: either install it (`pip install -e .` from the
repository root) or run the notebook from a directory where `import palsparserpy`
resolves.

````python
import os
import palsparserpy as pp

lattice_dir = os.path.join("..", "lattice_files")
ex_file     = os.path.join(lattice_dir, "ex.pals.yaml")
expand_file = os.path.join(lattice_dir, "expand.pals.yaml")
````

## Reading a lattice from a YAML file

Use `tree = pp.parse_file(filename)` to read any YAML file. To read a PALS file
*with* lattice expansion, use `pp.parse_and_expand_pals` instead.

````python
tree = pp.parse_file(ex_file)
````

## Printing a tree

To print a tree to the console, use `pp.to_yaml_string(tree)` -- or just
`print(tree)`, which does the same thing.

````python
print(pp.to_yaml_string(tree))
````

## Type checking

The root node of `ex.pals.yaml` is the `PALS` map, so `pp.is_map(tree)` is
`True`. The lattice contents live under the `facility` node of the `PALS` root,
which is a sequence.

````python
facility = tree["PALS"]["facility"]
pp.is_sequence(facility)
````

## Accessing a sequence

Elements in a sequence may be accessed by their index, counting from zero.

````python
seq1 = facility[0]
print(pp.to_yaml_string(seq1))
````

## Accessing a map

Elements in a map may be accessed by their key.

````python
map1 = seq1["constants"]["a_const"]
print(pp.to_yaml_string(map1))
````

## Finding a node's parent

Use `pp.get_parent(node)` to move up the tree. It returns the parent `YAMLNode`,
or raises if `node` is the root (which has no parent). Here the parent of the
`a_const` node above is the `constants` map that contains it.

````python
parent_node = pp.get_parent(map1)
print(pp.to_yaml_string(parent_node))
````

## Adding new elements

Add a new sequence element to the `facility` containing `new_map: {apples: 5}`.

````python
new_map_entry = facility.add_map()
map_node = new_map_entry.add_map(key="new_map")
map_node.add_scalar("5", key="apples")
````

Add another sequence element to the `facility` containing a `magnet_list`:

```yaml
- magnet_list:
    - magnet1
    - magnet2
```

````python
magnets_entry = facility.add_map()
sequence = magnets_entry.add_sequence(key="magnet_list")
sequence.add_scalar("magnet1")
sequence.add_scalar("magnet2", index=0)
````

## Writing the tree to a file

Use `pp.write_yaml(tree, filename)` to write the edited tree to a file.

````python
pp.write_yaml(tree, expand_file)
print("Wrote tree to 'expand.pals.yaml'")
````

## The final modified tree

````python
print(pp.to_yaml_string(tree))
````
